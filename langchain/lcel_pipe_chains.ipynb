```python
import os
```

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
```

```python
%load_ext dotenv
%dotenv
openai_api_key = os.getenv("api_key")
```

```python
RunnablePassthrough().invoke([1,2,3])
```

```python
chat_template_tools = ChatPromptTemplate.from_template(''' What are the five most important tools for a {job title} needs? Answer only by listing the tools ''')

chat_template_strategy = ChatPromptTemplate.from_template(''' Considering the tools provided, develop a strategy for effective learning and mastering them: {tools} ''')
```

```python
chat_template_tools
```

```python
chat = ChatOpenAI(model_name="gpt-4", 
                  seed=365,
                  temperature=0,
                  max_tokens=100,
                api_key=openai_api_key)
```

```python
string_parser = StrOutputParser()
```

```python
chain_tools = chat_template_tools | chat | string_parser | {"tools":RunnablePassthrough()}
chain_strategy = chat_template_strategy | chat | string_parser
```

```python
print(chain_tools.invoke({"job title": "data scientist"}))
```

```python
print(chain_strategy.invoke({"tools": "Python, SQL, Machine Learning, Data Visualization, Statistics"}))
```

```python
chain_combined = chain_tools | chain_strategy
```

```python
print(chain_combined.invoke({"job title": "data scientist"}))
```
