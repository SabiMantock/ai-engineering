### String Output Parser

```python
import os
from dotenv import load_dotenv
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_classic.output_parsers import DatetimeOutputParser
```

```python
load_dotenv()
openai_api_key = os.getenv("api_key")
```

```python
chat = ChatOpenAI(model_name="gpt-4o", seed=365, api_key=openai_api_key, temperature=0, max_tokens=100)
```

```python
message_h = HumanMessage(content= f"' When was the Danish poet Piet Hein born? {DatetimeOutputParser().get_format_instructions()} '")
```

```python
response = chat.invoke([message_h])
print(response.content)
```

```python
datetime_output_parser = DatetimeOutputParser()
```

```python
response_parsed = datetime_output_parser.invoke(response)
```
