### String Output Parser

```python
import os
from dotenv import load_dotenv
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import StrOutputParser
```

```python
load_dotenv()
openai_api_key = os.getenv("api_key")
```

```python
chat = ChatOpenAI(model_name="gpt-4o", seed=365, api_key=openai_api_key, temperature=0, max_tokens=100)
```

```python
message_h = HumanMessage(content="Can you give me an interesting fact i probably dont know about?")
```

```python
response = chat.invoke([message_h])
```

```python
response
```

```python
str_output_parser = StrOutputParser()
```

```python
response_parsed = str_output_parser.invoke(response)
response_parsed
```
