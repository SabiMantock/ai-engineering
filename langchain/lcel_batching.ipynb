# Batching

```python
import os
```

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
%load_ext dotenv
%dotenv
openai_api_key = os.getenv("api_key")
```

```python
chat_template = ChatPromptTemplate.from_messages([('human', "I've recently adopted a {pet} which is a breed. Could you suggest several training tips? \n")])
```

```python
chat = ChatOpenAI(model_name="gpt-4", temperature=0, api_key=openai_api_key, seed=365, max_tokens=100)
```

```python
chain = chat_template | chat
```

```python
chain.invoke({"pet": "dog", "breed": "sheperd"})
```

```python
%%time
chain.batch([{"pet": "dog", "breed": "sheperd"}, {"pet": "dragon", "breed": "night fury"}])
```

```python
%%time
chain.invoke({"pet": "dog", "breed": "sheperd"})
```

```python
%%time
chain.invoke({"pet": "dragon", "breed": "night fury"})
```

```python
response = chain.stream({"pet": "dragon", "breed": "night fury"})
```

```python
next(response)
```

```python
for i in response:
    print(i.content, end = "")
```
